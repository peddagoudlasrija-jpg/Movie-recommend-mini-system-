```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
movies = {
    "title": [
        "Avengers: Endgame",
        "Spider-Man: No Way Home",
        "Iron Man",
        "The Dark Knight",
        "Inception",
        "Interstellar",
        "Titanic",
        "The Notebook",
        "Jurassic Park",
        "The Lion King",
        "Frozen",
        "Toy Story",
        "Finding Nemo",
        "Avatar",
        "Black Panther"
    ],

    "genre": [
        "Action Adventure Science Fiction",
        "Action Adventure Science Fiction",
        "Action Science Fiction",
        "Action Crime Drama",
        "Science Fiction Thriller",
        "Science Fiction Drama Adventure",
        "Romance Drama",
        "Romance Drama",
        "Science Fiction Adventure",
        "Animation Adventure Drama",
        "Animation Adventure Fantasy",
        "Animation Comedy Adventure",
        "Animation Adventure Comedy",
        "Science Fiction Adventure",
        "Action Adventure Fantasy"
    ],

    "description": [
        "Superheroes fight to save the world from a powerful enemy",
        "Spider-Man faces villains from different universes",
        "A billionaire creates a powerful armored superhero suit",
        "A superhero fights a dangerous criminal in Gotham City",
        "A thief enters dreams to complete a difficult mission",
        "Astronauts travel through space searching for a new home",
        "A love story between two people on a famous ship",
        "A romantic story about two young people and their love",
        "Scientists discover dinosaurs and struggle to survive",
        "A young lion learns to become the king of the jungle",
        "A princess with magical powers tries to save her kingdom",
        "Toys come alive and experience exciting adventures",
        "A fish searches across the ocean for his lost son",
        "A human explores an alien planet and its civilization",
        "A superhero protects his country using advanced technology"
    ]
}

df = pd.DataFrame(movies)
df
```

```python
df["features"] = df["genre"] + " " + df["description"]

print("Movie Dataset:")
display(df[["title", "genre", "description"]])
```

```python
tfidf = TfidfVectorizer(stop_words="english")

tfidf_matrix = tfidf.fit_transform(df["features"])

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
```

```python
similarity = cosine_similarity(tfidf_matrix)

print("Similarity matrix created successfully!")
```

```python
def recommend_movies(movie_name, number_of_movies=5):

    movie_name = movie_name.lower()

    matches = df[df["title"].str.lower() == movie_name]

    if matches.empty:
        print("Movie not found in the dataset.")
        print("\nAvailable movies:")
        for movie in df["title"]:
            print("-", movie)
        return

    index = matches.index[0]

    similarity_scores = list(enumerate(similarity[index]))

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    recommendations = []

    for i, score in similarity_scores[1:number_of_movies+1]:
        recommendations.append({
            "Movie": df.iloc[i]["title"],
            "Genre": df.iloc[i]["genre"],
            "Similarity Score": round(score, 3)
        })

    result = pd.DataFrame(recommendations)

    print(f"\nMovies recommended for: {df.iloc[index]['title']}")
    display(result)
```

```python
recommend_movies("Avengers: Endgame")
```

```python
recommend_movies("Avengers: Endgame")
```

```python
recommend_movies("Titanic")
```
